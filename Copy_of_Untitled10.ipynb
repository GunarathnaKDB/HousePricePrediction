```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
from sklearn.datasets import fetch_california_housing
housing = fetch_california_housing()
```

```python
print(housing.DESCR)
```

```python
dataset=pd.DataFrame(housing.data,columns=housing.feature_names)
```

```python
dataset.head()
```

```python
dataset['price']=housing.target
```

```python
dataset.info()
```

**Summerizing the data(mean,median likewise)**

```python
dataset.describe()
```

**Check the missing values**

**Exploring data analysis**

**Plot part eka**

```python
plt.scatter(dataset['MedInc'],dataset['price'])
plt.xlabel('MedInc')
plt.ylabel('price')
```

```python
import seaborn as sns
sns.regplot(x=dataset['MedInc'],y=dataset['price'])
```

**Independent and Dependent features**

```python
# dataset එකෙහි 'price' column එක නොමැති නම් එය නැවත එකතු කරමු
if 'price' not in dataset.columns:
    dataset['price'] = housing.target

# price column එක ඉවත් කර ඉතිරි සියලුම columns (8 ක්) x ලෙස ලබා ගනිමු
x = dataset.drop(columns=['price'])
# price column එක පමණක් target (y) ලෙස ලබා ගනිමු
y = dataset['price']
```

**Training,Tesing split**

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
x_train
```

```python
x_test
```

**standardize the dataset**

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
x_train
```

```python
x_test
```

**#Model Training**

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
```

**Print the Coefficients and the intercepts**

```python
print(lr.coef_)
```

```python
print(lr.intercept_)
```

**Prediction  with TestData**

```python
reg_pred=lr.predict(x_test)
```

```python
reg_pred
```

**Plot a Scatter plot for the prediction**

```python
plt.scatter(y_test,reg_pred)
```

**Calculate the residuals**

```python
residuals=y_test-reg_pred
```

```python
residuals
```

**Plot the residuals using seaborn**

```python
sns.displot(residuals,kind='kde')
```

**Scatter plot with respect to prediction and residuals**

```python
plt.scatter(reg_pred,residuals)
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error

print(mean_absolute_error(y_test,reg_pred))
print(mean_squared_error(y_test,reg_pred))
print(np.sqrt(mean_squared_error(y_test,reg_pred)))
```

**R Square and adjusted R square**

```python
from sklearn.metrics import r2_score
score=r2_score(y_test,reg_pred)
print(score)
```
